## 7. Verify Training Results

```python
import os
import json

# Check if checkpoints were created
checkpoint_dir = "assets/checkpoints/weather_chpts/forecaster"
results_dir = "assets/results/weather/forecaster"

print("\n" + "="*60)
print("Training Results Summary")
print("="*60 + "\n")

models = ["timesnet", "gru", "dlinear"]

for model in models:
    checkpoint_file = f"chpt_weather_96_96_{model}_S.pth"
    history_file = f"history_weather_96_96_{model}_S.json"
    
    checkpoint_path = os.path.join(checkpoint_dir, checkpoint_file)
    history_path = os.path.join(results_dir, history_file)
    
    print(f"\n{model.upper()}:")
    print(f"  Checkpoint: {'✓' if os.path.exists(checkpoint_path) else '✗'} {checkpoint_path}")
    print(f"  History:    {'✓' if os.path.exists(history_path) else '✗'} {history_path}")
    
    if os.path.exists(history_path):
        with open(history_path, 'r') as f:
            history = json.load(f)
        print(f"  Epochs completed: {len(history['train_loss'])}")
        print(f"  Final train loss: {history['train_loss'][-1]:.5f}")
        print(f"  Final val loss:   {history['val_loss'][-1]:.5f}")
        print(f"  Final test loss:  {history['test_loss'][-1]:.5f}")
```

## 8. Display Training Curves

```python
from IPython.display import Image, display
import os

print("\n" + "="*60)
print("Training Curves")
print("="*60 + "\n")

for model in models:
    curve_path = f"assets/figures/weather/forecaster/{model}/{model}_training_curves.png"
    if os.path.exists(curve_path):
        print(f"\n{model.upper()} Training Curves:")
        display(Image(filename=curve_path))
    else:
        print(f"\n{model.upper()}: Training curves not found")
```
